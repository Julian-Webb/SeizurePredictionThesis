```python
from config import Paths
from pathlib import Path
import shutil
```

```python
# copy everything but the edf_data folders
src = Path('/data/home/webb/d/UNEEG')
dst = Path('/data/home/webb/d/UNEEG_cleaned')

def ignore_edf_data(dir, names):
    return [name for name in names if name == "edf_data"]

shutil.copytree(src, dst, dirs_exist_ok=True, ignore=ignore_edf_data)
```

```python
PATHS = Paths('/data/home/webb/d/UNEEG_cleaned')
PATHS
```

```python
# remove the svg segment plots (they're large) - keep the png ones
for pdir in PATHS.patient_dirs():
    segs_plot = pdir / 'segments_plot.svg'
    print(segs_plot)
    segs_plot.unlink(missing_ok=True)
```

```python
# unhide all files
for path in PATHS.rglob("*"):
    name = path.name
    if name.startswith(".") and name not in (".", ".."):
        new_name = name.removeprefix(".")
        new_path = path.with_name(new_name)
        print(f"{path}\n{new_path}\n")
        path.rename(new_path)
```

```python
# remove fake patients
for pdir in PATHS.patient_dirs():
    if 'FAKE' in pdir.name:
        print(pdir)
        shutil.rmtree(pdir)
```
