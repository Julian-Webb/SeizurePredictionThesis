```python
import numpy as np
import matplotlib.pyplot as plt


def get_safe_ci(boot_angles, obs_mean, ci_level=95):
    centered = wrap_angle(boot_angles - obs_mean)
    lower_p = (100 - ci_level) / 2.0
    l_c, u_c = np.percentile(centered, [lower_p, 100 - lower_p])
    return wrap_angle(l_c + obs_mean), wrap_angle(u_c + obs_mean)


def wrap_angle(angle):
    """Bounds an angle between -pi and pi."""
    return (angle + np.pi) % (2 * np.pi) - np.pi


def circular_mean(angles):
    """Calculates the circular mean."""
    S = np.sum(np.sin(angles))
    C = np.sum(np.cos(angles))
    return np.arctan2(S, C)


def permutation_test(phases1, phases2, n_iters=5000, ci_level=95, seed=42):
    """
    Bootstrap and permutation

    Parameters
    ----------
    phases1, phases2
        The phases of the two events in radians
    n_iters

    ci_level
    seed

    Returns
    -------

    """
    np.random.seed(seed)
    ph1 = np.asarray(phases1)
    ph2 = np.asarray(phases2)
    n1, n2 = len(ph1), len(ph2)

    # Calculate true observed means and difference
    obs_mean1 = circular_mean(ph1)
    obs_mean2 = circular_mean(ph2)
    obs_diff = wrap_angle(obs_mean1 - obs_mean2)

    # =========================================================
    #  THE BOOTSTRAP (Calculates the physical 95% CI)
    # =========================================================
    idxs1 = np.random.randint(0, n1, size=(n_iters, n1))
    idxs2 = np.random.randint(0, n2, size=(n_iters, n2))

    S1, C1 = np.sum(np.sin(ph1[idxs1]), axis=1), np.sum(np.cos(ph1[idxs1]), axis=1)
    S2, C2 = np.sum(np.sin(ph2[idxs2]), axis=1), np.sum(np.cos(ph2[idxs2]), axis=1)

    boot_means1 = np.arctan2(S1, C1)
    boot_means2 = np.arctan2(S2, C2)
    boot_diffs = wrap_angle(boot_means1 - boot_means2)

    # Calculate CIs for individual groups (for polar plot)
    lower1, upper1 = get_safe_ci(boot_means1, obs_mean1, ci_level=ci_level)
    lower2, upper2 = get_safe_ci(boot_means2, obs_mean2, ci_level=ci_level)

    # Calculate CI for the difference (for histogram/reporting)
    ci_lower, ci_upper = get_safe_ci(boot_diffs, obs_diff, ci_level=ci_level)

    # Permutation for p_val
    pool = np.concatenate([ph1, ph2])
    n_total = len(pool)

    shuffled_idxs = np.random.rand(n_iters, n_total).argsort(axis=1)
    shuffled_pool = pool[shuffled_idxs]

    pseudo1 = shuffled_pool[:, :n1]
    pseudo2 = shuffled_pool[:, n1:]

    S_p1, C_p1 = np.sum(np.sin(pseudo1), axis=1), np.sum(np.cos(pseudo1), axis=1)
    S_p2, C_p2 = np.sum(np.sin(pseudo2), axis=1), np.sum(np.cos(pseudo2), axis=1)

    perm_diffs = wrap_angle(np.arctan2(S_p1, C_p1) - np.arctan2(S_p2, C_p2))
    p_value = (np.sum(np.abs(perm_diffs) >= np.abs(obs_diff)) + 1) / (n_iters + 1)

    return {
        "observed_diff": obs_diff,
        "ci_lower": ci_lower,
        "ci_upper": ci_upper,
        "p_value": p_value,
        "boot_diffs": boot_diffs,
        "perm_diffs": perm_diffs,
        "obs_mean1": obs_mean1,
        "lower1": lower1,
        "upper1": upper1,
        "obs_mean2": obs_mean2,
        "lower2": lower2,
        "upper2": upper2
    }
```

```python
import numpy as np
import matplotlib.pyplot as plt

def get_safe_ci(boot_angles, obs_mean, ci_level=95):
    centered = wrap_angle(boot_angles - obs_mean)
    lower_p = (100 - ci_level) / 2.0
    l_c, u_c = np.percentile(centered, [lower_p, 100 - lower_p])
    return wrap_angle(l_c + obs_mean), wrap_angle(u_c + obs_mean)

def wrap_angle(angle):
    """Bounds an angle between -pi and pi."""
    return (angle + np.pi) % (2 * np.pi) - np.pi

def circular_mean(angles):
    """Calculates the circular mean."""
    S = np.sum(np.sin(angles))
    C = np.sum(np.cos(angles))
    return np.arctan2(S, C)

def permutation_test(seizures, fps, n_iters=5000, ci_level=95, seed=42):
    """
    Bootstrap + permutation
    """
    np.random.seed(seed)
    szr = np.asarray(seizures)
    fp = np.asarray(fps)
    n_szr, n_fp = len(szr), len(fp)

    # Calculate true observed means and difference
    obs_m_szr = circular_mean(szr)
    obs_m_fp = circular_mean(fp)
    obs_diff = wrap_angle(obs_m_szr - obs_m_fp)

    # =========================================================
    #  THE BOOTSTRAP (Calculates the physical 95% CI)
    # =========================================================
    idx_szr = np.random.randint(0, n_szr, size=(n_iters, n_szr))
    idx_fp = np.random.randint(0, n_fp, size=(n_iters, n_fp))

    S_szr, C_szr = np.sum(np.sin(szr[idx_szr]), axis=1), np.sum(np.cos(szr[idx_szr]), axis=1)
    S_fp, C_fp = np.sum(np.sin(fp[idx_fp]), axis=1), np.sum(np.cos(fp[idx_fp]), axis=1)

    boot_szr_means = np.arctan2(S_szr, C_szr)
    boot_fp_means = np.arctan2(S_fp, C_fp)
    boot_diffs = wrap_angle(boot_szr_means - boot_fp_means)

    # Calculate CIs for individual groups (for polar plot)
    szr_lower, szr_upper = get_safe_ci(boot_szr_means, obs_m_szr, ci_level=ci_level)
    fp_lower, fp_upper = get_safe_ci(boot_fp_means, obs_m_fp, ci_level=ci_level)

    # Calculate CI for the difference (for histogram/reporting)
    ci_lower, ci_upper = get_safe_ci(boot_diffs, obs_diff, ci_level=ci_level)

    # =========================================================
    # PERMUTATION TEST
    # =========================================================
    R_szr = np.sqrt(np.sum(np.sin(szr))**2 + np.sum(np.cos(szr))**2) / n_szr # same as abs(np.sum(np.exp(1j * event_phases)) / n_events)-->PLV
    R_fp = np.sqrt(np.sum(np.sin(fp))**2 + np.sum(np.cos(fp))**2) / n_fp

    eps = 1e-8 # buffer to prevent zero div.
    var_szr = (1.0 - R_szr) / (n_szr * (R_szr**2 + eps))
    var_fp = (1.0 - R_fp) / (n_fp * (R_fp**2 + eps))

    obs_T = np.abs(obs_diff) / np.sqrt(var_szr + var_fp) # similar to Welch's t-test

    pool = np.concatenate([szr, fp])
    n_total = len(pool)
    shuffled_indices = np.random.rand(n_iters, n_total).argsort(axis=1)
    shuffled_pool = pool[shuffled_indices]

    pseudo_szr = shuffled_pool[:, :n_szr]
    pseudo_fp = shuffled_pool[:, n_szr:]

    S_p1, C_p1 = np.sum(np.sin(pseudo_szr), axis=1), np.sum(np.cos(pseudo_szr), axis=1)
    S_p2, C_p2 = np.sum(np.sin(pseudo_fp), axis=1), np.sum(np.cos(pseudo_fp), axis=1)

    perm_diffs = wrap_angle(np.arctan2(S_p1, C_p1) - np.arctan2(S_p2, C_p2))

    R_p1 = np.sqrt(S_p1**2 + C_p1**2) / n_szr
    R_p2 = np.sqrt(S_p2**2 + C_p2**2) / n_fp

    var_p1 = (1.0 - R_p1) / (n_szr * (R_p1**2 + eps))
    var_p2 = (1.0 - R_p2) / (n_fp * (R_p2**2 + eps))

    perm_T = np.abs(perm_diffs) / np.sqrt(var_p1 + var_p2)
    p_value = (np.sum(perm_T >= obs_T) + 1) / (n_iters + 1)

    return {
        "observed_diff": obs_diff,
        "ci_lower": ci_lower,
        "ci_upper": ci_upper,
        "p_value": p_value,
        "boot_diffs": boot_diffs,
        "perm_diffs": perm_diffs,
        "obs_m_szr": obs_m_szr,
        "szr_lower": szr_lower,
        "szr_upper": szr_upper,
        "obs_m_fp": obs_m_fp,
        "fp_lower": fp_lower,
        "fp_upper": fp_upper
    }
```
