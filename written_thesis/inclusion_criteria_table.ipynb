```python
from pathlib import Path
import pandas as pd

from config import PATHS
from written_thesis.helpers import rename_patient
```

```python
# Paths
# on server:
in_path = PATHS.patient_info_exact.pickle
out_path = Path("~/thesis_data/tables/patient_inclusion.tex").expanduser()

# local:
in_path = Path("~/thesis_files/patient_info/.patient_info_exact.pkl").expanduser()
out_path = Path('~/Developer/MastersThesis/src/tables/patient_inclusion.tex').expanduser()
```

```python
orig = pd.read_pickle(in_path)

# Keep only real patients (exclude names containing "FAKE")
patient_names = orig.index.get_level_values("patient")
orig = orig[~patient_names.str.contains("FAKE", regex=False)]
orig.head()
```

```python
patient_level = orig.index.names.index("patient")
renamed_patient_levels = orig.index.levels[patient_level].map(rename_patient)
orig.index = orig.index.set_levels(renamed_patient_levels, level="patient")

orig
```

```python
no_edfs_mask = orig['recordings_start'].isna()
no_edfs = orig[no_edfs_mask]
has_edfs = orig[~no_edfs_mask].copy()
print('No EDFs, excluding: ', no_edfs.index.get_level_values('patient').to_list())
no_edfs
```

```python
has_edfs
```

```python
# Make it pretty
t = has_edfs.copy()
t = t.sort_index().reset_index()
# t = t.assign(valid=t.pop('valid'))  # move valid to the end
t = t.drop(columns=['dataset', 'recordings_start', 'recordings_end', 'duration_not_recorded'])
for col in ['duration_recorded', 'timespan']:
    t[col] = t[col].dt.days  # just show days (not hours)
for col in ['valid', 'enough_valid_seizures', 'valid_ratio_recorded']:
    t[col] = t[col].map({True: r'$\checkmark$', False: r'$\times$'}).astype(str)
t['ratio_recorded'] = (t['ratio_recorded'] * 100).round().astype(int)

assert t.columns.to_list() == ['patient', 'valid', 'total_seizures', 'valid_seizures', 'enough_valid_seizures',
                               'timespan', 'duration_recorded', 'ratio_recorded', 'valid_ratio_recorded']
# Grouped headers
t.columns = pd.MultiIndex.from_tuples([
    ('Patient', 'ID'),
    ('Patient', 'Incl.'),
    ('Seizures', 'Total'),
    ('Seizures', 'Valid'),
    ('Seizures', 'Met'),
    ('Timespan (days)', 'Monitoring'),
    ('Timespan (days)', 'Recorded'),
    ('Timespan (days)', 'Coverage'),
    ('Timespan (days)', 'Met'),
])

t
```

```python
latex = t.to_latex(
    index=False,
    escape=False,
    formatters={("Timespan (days)", "Coverage"): lambda x: r"\SI{" + str(x) + r"}{\percent}"},
    multicolumn=True,
    multicolumn_format="|c|",
    column_format="|l r|r r r|r r r r|",  # or no vertical bars if using pure booktabs style
)

# Add row color to header rows
lines = latex.split("\n")
in_header = False
new_lines = []
for i, line in enumerate(lines):
    if r"\toprule" in line:
        in_header = True
        new_lines.append(line)
    elif in_header and line.strip().endswith(r"\\"):
        # This is a header row, add \rowcolor
        new_lines.append(r"\rowcolor{gray!25} " + line)
    elif in_header and r"\midrule" in line and i > 0:
        # \midrule marks end of headers
        new_lines.append(line)
        in_header = False
    else:
        new_lines.append(line)
latex = "\n".join(new_lines)

# Replace booktabs rules with standard hlines
latex = latex.replace(r"\toprule", r"\hline").replace(r"\midrule", r"\hline").replace(r"\bottomrule", r"\hline")

# Fix thicker middle vertical lines
latex = (latex
         .replace(r'\multicolumn{3}{|c|}{Seizures}', r'\multicolumn{3}{c|}{Seizures}')
         .replace(r'\multicolumn{4}{|c|}{Timespan (days)}', r'\multicolumn{4}{c|}{Timespan (days)}'))

out_path.write_text(latex, encoding="utf-8")
```
