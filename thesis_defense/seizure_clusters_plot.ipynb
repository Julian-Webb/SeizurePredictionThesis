```python
from config import PATHS
import pandas as pd
import matplotlib.pyplot as plt

from written_thesis.helpers import rename_patient
```

```python
pdir = PATHS.patient_dirs(include_fake_ptnts=False)[3]
pdir
```

```python
szrs = pd.read_pickle(pdir.valid_szr_starts_file.pickle)['start_mtz']
partition = pd.read_pickle(pdir.dataset_partition.pickle)
test_start = partition.loc['test', 'start_mtz']
```

```python
fig, ax = plt.subplots(figsize=(15, 1))
ax.vlines(szrs, ymin=0, ymax=1, linewidth=0.5, color='red', label='seizures')
ax.vlines(test_start, ymin=-.1, ymax=1.1, linewidth=2, color='green', linestyles='--', label='test set start', clip_on=False)

ax.set_ylim(0, 1)
plt.yticks([])

ax.set_title(f'Patient {rename_patient(pdir.name)} seizures')
fig.legend(loc='center right')
plt.show()
```
