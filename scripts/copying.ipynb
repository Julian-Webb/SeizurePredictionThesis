Various scripts to copy important information to one place for better viewing

```python
# Copy metrics plots to single dir
import os
import shutil
from pathlib import Path

from config import Paths

datasets = [
    {'abbreviated': 'former', 'full': 'UNEEG_20260409_models_before_fair_partitioning'},
    {'abbreviated': 'present', 'full': 'UNEEG'},
]

home = Path('/data/home/webb/')
os.chdir(home)

dst_root = Path('metric_plots')
if dst_root.exists():
    shutil.rmtree(dst_root)

for dataset in datasets:
    dataset_paths = Paths('d', dataset['full'])

    for pdir in dataset_paths.patient_dirs():
        if 'FAKE' in pdir.name:
            continue

        for split in ['train', 'test']:
            src = pdir.model_eval_subdir(split).metrics_plot
            dst = dst_root / split / f"{pdir.name}_{split}_{dataset['abbreviated']}_metrics_plot.png"
            dst.parent.mkdir(parents=True, exist_ok=True)
            print(src)
            print(dst)
            print()
            shutil.copy2(src, dst)
```

```python
# Copy Cycle Extraction plots to single dir
from config import PATHS
from pathlib import Path
import shutil

dst_root = Path('/data/pool/webb/UNEEG_Cycle_Extraction_Plots')
if dst_root.exists():
    shutil.rmtree(dst_root)

for pdir in PATHS.patient_dirs():
    if 'FAKE' in pdir.name:
        continue

    dst_pdir = dst_root / pdir.name

    for src_dir in [pdir.filtered_feature_plots_dir, pdir.circular_histograms_dir]:
        dst_dir = dst_pdir / src_dir.name
        shutil.copytree(src_dir, dst_dir, copy_function=shutil.copy2)

# Make readme
readme_path = dst_root / 'README.txt'
readme_lines = [
    "Filtered features plots and circular histograms for all patients",
    "Just for TEST data",
]
readme_path.write_text('\n'.join(readme_lines), encoding='utf-8')
```
