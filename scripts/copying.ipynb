Various scripts to copy important information to one place for better viewing

```python
# Copy metrics plots to single dir
import os
import shutil
from pathlib import Path

from config import Paths

datasets = [
    {'abbreviated': 'former', 'full': 'UNEEG_20260409_models_before_fair_partitioning'},
    {'abbreviated': 'present', 'full': 'UNEEG'},
]

home = Path('/data/home/webb/')
os.chdir(home)

dst_root = Path('metric_plots')
if dst_root.exists():
    shutil.rmtree(dst_root)

for dataset in datasets:
    dataset_paths = Paths('d', dataset['full'])

    for pdir in dataset_paths.patient_dirs():
        if 'FAKE' in pdir.name:
            continue

        for split in ['train', 'test']:
            src = pdir.model_eval_subdir(split).metrics_plot
            dst = dst_root / split / f"{pdir.name}_{split}_{dataset['abbreviated']}_metrics_plot.png"
            dst.parent.mkdir(parents=True, exist_ok=True)
            print(src)
            print(dst)
            print()
            shutil.copy2(src, dst)
```

```python
# Copy Cycle Extraction plots to thesis figures dir
import shutil
from config import PATHS
from written_thesis.helpers import THESIS_FIGS_DIR, rename_patient

for pdir in PATHS.patient_dirs(include_fake_ptnts=False):
    for folder in [pdir.filtered_feature_plots_dir, pdir.phase_histograms_dir]:
        dst = THESIS_FIGS_DIR / folder.name / rename_patient(pdir.name)
        shutil.copytree(folder, dst, dirs_exist_ok=True)
```
